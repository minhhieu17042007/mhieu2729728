W6A20: Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1



```python
def so_dau_tien(n) : 
    if k in n : 
        return n.index(k)
    else : 
        return -1 
k = int(input())
n = list(map(int,input().split()))
print(so_dau_tien(n))
    
```

W6A28
Cho một dãy các số nguyên, kiểm tra xem nó là dãy tăng thực sự hay giảm
thực sự, 

```python
def day(n) : 
    Tang = False 
    Giam = False 
    for i in range(0,len(n)-1) : 
        if n[i] <= n[i+1] : 
            Tang = True 
        elif n[i] >= n[i+1] : 
            Giam = True

    if Tang:
        return "Tăng "
    elif Giam : 
        return "Giảm"
    else : 
        return "Vo danh"
n = list(map(int,input().split()))
print(day(n))
```

W6A24
Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí 

```python
def day(n) : 
    kq = []
    so_khac = []
    for so in n : 
        if so == 0 : 
            kq.append(so)
        else :
            so_khac.append(so)
    x = kq + so_khac 
    return x 
n = n = list(map(int,input().split()))
print(day(n))
```
