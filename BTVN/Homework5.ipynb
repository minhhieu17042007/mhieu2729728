W5A16 : Viết hàm kiểm tra một số có phải số Amstronhg hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

```python
def so_Amstronhg(x) : 
    so = 0 
    tong = 0 
    while x > 0 : 
        n = x % 10 
        so = so + 1 
        x = x //10 
        tong = tong + n**so
    if tong == x : 
        return True
    else : 
        return False 
    
x = int(input())
print(so_Amstronhg(x))    
```
