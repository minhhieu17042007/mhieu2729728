W5A17 : Viết hàm kiểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số
nguyên x

W5A23 : Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.


```python
import math
def quadratic_roots_count(a,b,c) :
    if a == 0 and b != 0 : 
        x = -c / b 
        return f"Phương trình có nghiệm duy nhất là : {x} "
    elif a == 0 and b == 0 : 
        if c == 0 : 
            return "Phương trình có vô số nghiệm "
        else : 
            return "Phương trình vô nghiệm"
    else :
        d = b**2 - 4*a*c 
        if d < 0 : 
            return "Phương trình vô nghiệm"
        elif d == 0 : 
            x = -b /2*a 
            return f"Phương trình có nghiệm duy nhất là : {x} "
        else : 
            x1 = (-b-math.sqrt(d)) / 2*a
            x2 = (-b+math.sqrt(d)) / 2*a
            if x1 < x2 : 
                return x1,x2
            else : 
                return x2,x1


a , b ,c = map(float,input().split())
print(quadratic_roots_count(a,b,c))
```

W5A24 : Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không

```python
def are_collinear(x1,y1,x2,y2,x3,y3) : 
    xAB= x2 - x1 
    yAB = y2 - y1
    xAC = x3 - x1 
    yAC = y3 - y1 
    if (xAB/xAC) == (yAB/yAC) : 
        return True 
    else : 
        return False 
x1 , y1 , x2 , y2 , x3 , y3 = map(int, input().split())
print(are_collinear(x1,y1,x2,y2,x3,y3))    
```

W5A28 : Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).

```python
import math
def goc(x1,y1,x2,y2,x3,y3):
    a = x1-x2
    b = y1-y2
    c = x3-x2
    d = y3-y2
    cosa = (a*c+b*d)/ (math.sqrt(a**2+b**2)*math.sqrt(c**2+d**2))
    radian = math.acos(cosa)
    do = math.degrees(radian)
    return do

x1 , y1 , x2 , y2 , x3 , y3 = map(int, input().split())
print(goc(x1,y1,x2,y2,x3,y3))
```

W5A29 : Viết hàm trả về giá trị lớn nhất và nhỏ nhất của ba số a,b,c dưới dạng list [max, min]

```python
def so(a,b,c) : 
    max = a 
    if b > a and b > c : 
        max = b 
    elif c > a and c > b : 
        max = c 
    
    min = a 
    if a > c and b > c : 
        min = c 
    elif a > b and c > b : 
        min = b 
    
    return [max,min]

a,b,c = map(int,input().split())
print(so(a,b,c))
    
```

W5A30: Viết hàm tính hệ số nhị thức C(n, k).

```python
def he_so_nhi_thuc(n,k) : 
    a = n - k 
    giai_thua_n = 1 
    giai_thua_k = 1
    giai_thua_a = 1 
    for i in range(1,n+1) : 
        giai_thua_n = giai_thua_n * i 
    for x in range(1,k+1) : 
        giai_thua_k = giai_thua_k * x 
    for b in range(1,a+1) : 
        giai_thua_a = giai_thua_a * b 
    
    return giai_thua_n / giai_thua_k * giai_thua_a
n,k = map(int,input().split())
print(he_so_nhi_thuc(n,k))
    
```
