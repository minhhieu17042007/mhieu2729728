```python

while True :
    n = int(input()) 
    if n > 0 : 
        print(f"{n} là một số nguyên dương")
        break   
    else : 
        print("Error")

if n < 2 : 
    print (f"{n} không phải số nguyên tố ")
else : 
    for i in range (2,n): 
        if n % i == 0 : 
          print (f"{n} là không phải số nguyên tố")
          break
    else :
        print (f"{n} là một số nguyên tố")
```

W4A8 : ReverseAddPalindrome]
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số
đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 


```python
def deverse(n) : 
    so = n 
    x= 0 
    while so > 0 : 
        a = so % 10 
        x = x*10 + a 
        so = so // 10 
    return x 
print(deverse(n))
```

W4A10 : "[LongestCollatzUpToN]
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất
và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số
nhỏ nhất.

```python

n = int(input())
max_l, res = 0, 0

for i in range(1, n + 1):
    curr, l = i, 1
    while curr > 1:
        curr = curr // 2 if curr % 2 == 0 else 3 * curr + 1
        l += 1
    
    if l > max_l:
        max_l, res = l, i

print(res, max_l)
```
