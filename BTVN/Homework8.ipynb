Xây dựng lớp Shape với phương thức area().
Xây dựng lớp Rectangle(w, h) và Circle(r) kế thừa từ Shape, ghi đè area().
Lấy π = 3.14.
Input
Dòng 1: loại hình (RECTANGLE hoặc CIRCLE)
Dòng 2: tham số tương ứng.
Output
In ra diện tích với 2 chữ số thập phân

```python
class Shape():
    def area(self):
        pass
class Rectangle(Shape) : 
    def __init__ (self,w,h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
class Circle(Shape) : 
    def __init__(self,r):
        self.r = r
    def area(self):
        return (self.r**2)*3.14
loai = input().strip()
dai = list(map(int,input().split()))
if loai == "Rectangle" : 
    w = float(dai[0])
    h = float(dai[1])
    hinh = Rectangle(w,h)
elif loai == "Circle" : 
    r = float(dai[0])
    hinh = Circle(r)
print(format(hinh.area()))


```

Xây dựng lớp Account(balance) với hai phương thức deposit(x) và get_balance().
Xây dựng lớp SavingAccount(balance, rate) kế thừa từ Account, thêm phương thức
add_interest() để cộng balance*rate vào số dư.
Input
balance rate
Output
Số dư sau khi cộng lãi, in 2 chữ số thập phân. 

```python
class Account():
    def __init__ (self,balance) : 
        self.balance = float(balance)
    def deposit(self,x):
        self.balance += x
    def get_balance(self):
        return self.balance
class SavingAccount(Account):
    def __init__ (self,balance,rate):
        super(). __init__(balance)
        self.rate = float(rate)
    def add_interest(self):
        interest =  self.balance * self.rate
        self.deposit(interest)
b,r = input().split()
n = SavingAccount(b,r)
n.add_interest()
print(f"{n.get_balance():.2f}")
    
```

W8A22: Yêu cầu
Xây dựng lớp Matrix2x2(a, b, c, d) biểu diễn ma trận:
[a b]
[c d]
Viết phương thức trace() trả về a + d và determinant() trả về ad - bc.
Input
Một dòng chứa 4 số nguyên a b c d.
Output
In ra trace và determinant trên cùng một dòng. 


```python
class Matrix2x2():
    def __init__ (self,a,b,c,d) :
        self.mtr = [
            [a,b],
            [c,d]
        ]
    def trace(self):
        return self.mtr[0][0] + self.mtr[1][1]
    def determinant(self):
        return self.mtr[0][0]*self.mtr[1][1]-self.mtr[0][1]*self.mtr[1][0]
a,b,c,d = map(int,input().split())
ma_tran = Matrix2x2(a,b,c,d)
ma_tran.trace() , ma_tran.determinant()
```

Viết phương thức sum_str() trả về chuỗi biểu diễn đa thức tổng dưới dạng kx + m.

a b c d

Chuỗi của đa thức P(x)+Q(x). 

```python
class PolynomialSum():
    def __init__ (self,a,b,c,d) : 
        self.a = a 
        self.b = b 
        self.c = c 
        self.d = d 
    def sum_str(self) : 
        k = self.a + self.c
        m = self.b + self.d 
        return f"{k}x + {m}"
a,b,c,d = map(int,input().split())
n = PolynomialSum(a,b,c,d)
print(n.sum_str())
```

W8A27 : Yêu cầu
Xây dựng lớp Product(name, price).
Xây dựng lớp DiscountProduct(name, price, percent) kế thừa từ Product, ghi đè phương thức
final_price() để trả về giá sau giảm.
Công thức: price * (100 - percent) / 100
Input
name price percent  

```python
class Product():
    def __init__ (self,name,price) : 
        self.name = name 
        self.price = float(price)
    def final_price(self):
        return self.price
class DiscountProduct(Product):
    def __init__ (self,name,price,percent) : 
        super().__init__(name,price)
        self.percent = float(percent)
    def final_price(self):
        n = self.price * (100 - self.percent)/100
        return round(n,2)
name,price,percent = input().split()
a = DiscountProduct(name,price,percent)
print(f"Tên sản phẩm: {a.name} | Giá cuối: {a.final_price()}")
```

Xây dựng lớp Counter có thuộc tính count ban đầu bằng 0 và các phương thức inc(), dec(),
value().
Sau đó thực hiện dãy thao tác trên đối tượng.
Input
Dòng 1: n
Dòng 2: n thao tác, mỗi thao tác là INC hoặc DEC, cách nhau bởi dấu cách.
Output
Giá trị cuối cùng của count. 


```python
class Counter():
    def __init__ (self,count):
        self.count = 0 
    def inc(self):
        self.count += 1 
    def dec(self): 
        self.count -= 1 
    def value(self) : 
        return self.count
n = int(input())
lenh = input().split()
a = Counter(0)
for i in range(n) : 
    for cmd in lenh : 
        if cmd == "INC" : 
            a.inc()
        elif cmd == "DEC" :
            a.dec()
print(a.value())
```

Xây dựng lớp Time(h, m, s) với phương thức to_seconds() đổi thời gian sang tổng số giây.
Input
h m s
Output
Tổng số giây

```python
class Time():
    def __init__(self,h,m,s) : 
        self.h = h 
        self.m = m 
        self.s = s 
    def to_seconds(self):
        n = self.h*3600 + self.m*60 + self.s 
        return n 
h,m,s = map(int,input().split())
n = Time(h,m,s)
print(n.to_seconds())
```

Xây dựng lớp Stack sử dụng danh sách để cài đặt ngăn xếp với các phương thức push(x),
pop(), top().
Dữ liệu đảm bảo mọi lệnh pop và top đều hợp lệ.
Input
Dòng 1: n
n dòng tiếp theo là một trong ba lệnh:
PUSH x
POP
TOP
Output
Mỗi lệnh TOP in ra giá trị ở đỉnh ngăn xếp trên một dòng. 

```python
class Stack():
    def __init__(self):
        self.lst = []
    def push(self,x) : 
        self.lst.append(x)
    def pop(self):
        self.lst.pop()
    def top(self):
        return self.lst[-1]
n = int(input())
a = Stack()
for i in range(n):
    lenh = input().strip()
    if lenh == "PUSH" : 
        x = input().split()
        for s in x : 
            a.push(int(s))
    elif lenh == "POP":
        a.pop()
    elif lenh == "TOP" :   
        print(a.top())
```
