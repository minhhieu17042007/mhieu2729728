W4A1 :SumOfNumBers
Viết chương trình Python sử dụng một trong 2 loại vòng lặp for hoặc while để
tính tống các số nguyên từ 1 đến n. (n <= 1000) 


```python
n = int(input())
tong = 0 
for i in range(1,n+1):
    tong = tong + i 
print(tong)
 
```

```python
tong = 0 
so = 1 
while so <= 6 : 
    tong = tong + so 
    so = so +1 
print(tong)
```

W4A2 : VIết chương trình nhập vào một số tới khi được một số nguyên dương thì thôi.
Kiểm tra số này có phải là số nguyên tố hay không? 

```python
n = int(input())
while True : 
    if n > 0 : 
        print(f"{n} là một số nguyên dương")
        break
    else : 
        print("Error")

if n < 2 : 
    print (f"{n} không phải số nguyên tố ")
else : 
    for i in range (2,n): 
        if n % i != 0 : 
           print (f"{n} là 1 số nguyên tố ")
           break
        else : 
           print ("Không phải số nguyên tố")
```

W4A3 : [Factorial]
Viết chương trình tính giai thừa của một số nguyên n cho trước (n!=1×2×…×n)
với 0 < n < 100. 

```python
n = int(input())
giai_thua = 1
for i in range(1,n+1): 
    giai_thua = giai_thua * i 
print (giai_thua)
```
